```python
from dotenv import load_dotenv
from crewai import LLM, Agent, Task, Crew

load_dotenv()

llm = LLM(
    model="gpt-4o-mini",temperature=0.7,)

# -----------------------------
# AGENTE 1: Investigador
# -----------------------------
# Propósito:
# Este agente es responsable de investigar y resumir
# las últimas tendencias y desarrollos en IA.
#
# Piensa en este agente como un analista de investigación
# que recopila información importante.
researcher = Agent(  
    role="Researcher",  
    goal="Summarize the latest trends in AI for Software Development.",  
    backstory="You are an AI assistant tasked with distilling complex topics into clear insights.",  
    llm=llm,
    verbose=True
)

print(f"Agente 1: {researcher.role} - {researcher.goal}")

# -----------------------------
# AGENTE 2: Escritor
# -----------------------------
# Propósito:
# Este agente toma los hallazgos de la investigación
# y los convierte en un informe legible.
#
# Piensa en este agente como un escritor técnico
# que transforma información en bruto en
# contenido estructurado y atractivo.
writer = Agent(  
    role="Writer",  
    goal="Generate a concise weekly report based on researcher's findings.",  
    backstory="You are a skilled writer who transforms data into actionable insights.",  
    llm=llm,
    verbose=True
)  


# -----------------------------
# TAREA 1: Tarea de Investigación
# -----------------------------
# Propósito:
# Asigna al agente Investigador el trabajo de
# analizar los desarrollos recientes en IA.
#
# Resultado Esperado:
# Un resumen en viñetas de las tendencias importantes en IA.
research_task = Task(  
    description=("Analyze recent AI developments and summarize key points."),  
    expected_output="A bullet-point summary of the latest AI trends.",
    agent=researcher 
)  

# -----------------------------
# TAREA 2: Tarea de Escritura
# -----------------------------
# Propósito:
# Asigna al agente Escritor la responsabilidad
# de crear un informe completo usando la
# investigación generada en la Tarea 1.
#
# Contexto:
# El escritor recibe la salida generada
# por la tarea de investigación como contexto de entrada.
#
# Resultado Esperado:
# Un informe semanal de IA pulido de 500 palabras.
writing_task = Task(  
    description="Write a 500-word report summarizing the research.",  
    expected_output="A bullet-point summary of the latest AI trends.",
    context = [research_task],
    agent=writer  
)  

# Crear el crew y ejecutar las tareas
crew = Crew(agents=[researcher, writer], tasks=[research_task, writing_task])  
result = await crew.kickoff_async()

from IPython.display import Markdown
Markdown(result.raw)
```
