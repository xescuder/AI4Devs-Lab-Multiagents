# Un planificador de viajes, 7 arquitecturas distintas

Notebook con los 7 ejemplos de patrones multiagente (CrewAI) aplicados al mismo problema: **planificar un viaje de 4 días a Lisboa para 2 personas con presupuesto de 1200€**. Lo único que cambia entre celdas es la arquitectura — no el dominio.

Especialistas que reaparecen en casi todos los ejemplos:

- **Vuelos** — busca y compara opciones de vuelo
- **Hoteles** — busca alojamiento dentro de presupuesto
- **Actividades** — propone un plan de actividades
- **Transporte** — traslados y movilidad local


## Importante: no los 7 son "multiagente" en sentido fuerte

Usar varios agentes en un ejemplo no lo convierte automáticamente en un sistema multiagente. En un **workflow**, el camino lo decide tu código de antemano; en un sistema **agéntico/multiagente**, quien decide a quién llamar y en qué orden es el propio LLM, en tiempo de ejecución.

Con ese criterio:

| # | Patrón | ¿Quién decide el camino? | Clasificación real |
|---|---|---|---|
| 1 | Chaining | Yo, en el orden de `tasks=[...]` y los `context=[...]` | **Workflow** |
| 2 | Routing | Yo, con una función Python (`if/elif`), ni siquiera un LLM | **Routing por código** |
| 3 | Parallelization | Yo, fijando qué 4 tasks van en paralelo y cuál las combina | **Workflow paralelo** |
| 4 | Orchestrator-Workers | El **manager LLM** (`Process.hierarchical`), en tiempo real | **Multiagente real** |
| 5 | Evaluator-Optimizer | Yo decido el bucle; 1 agente genera, una función Python evalúa | **Ni workflow multi-paso ni multiagente** |
| 6 | Jerárquico | Yo, en Python, pasando resultados a mano entre crews | **Workflow que orquesta varios crews** |
| 7 | Democrático | Los agentes votan; el escrutinio sigue instrucciones fijas mías | **Multiagente débil** |

**Solo el patrón 4 (Orchestrator-Workers) es multiagente en sentido estricto.** Los demás son workflows bien diseñados que usan varios agentes como piezas dentro de una estructura que tú ya definiste.

## 1. Prompt Chaining — itinerario paso a paso

> **Clasificación: Workflow.** El orden vuelos→hotel→actividades lo fijo yo en el código; ningún agente decide llamar a otro.

**Por qué este patrón aquí:** planificar un viaje tiene un orden natural y fijo — primero sabes cuándo vuelas, luego reservas el hotel para esas fechas, luego encajas actividades en los días que queden. Cada paso necesita el resultado exacto del anterior.


```python
from crewai import Agent, Task, Crew, Process

destino, dias, personas, presupuesto = "Lisboa", 4, 2, 1200

vuelos_agent = Agent(
    role="Especialista en Vuelos",
    goal="Encontrar las mejores opciones de vuelo dentro de presupuesto",
    backstory="Conoces tarifas y aerolíneas para rutas europeas.",
)
vuelos_task = Task(
    description=f"Propón opciones de vuelo a {destino} para {personas} personas, {dias} días. Presupuesto total del viaje: {presupuesto}€.",
    expected_output="2-3 opciones de vuelo con precio aproximado y fechas, y cuánto presupuesto queda tras los vuelos.",
    agent=vuelos_agent,
)

hoteles_agent = Agent(
    role="Especialista en Hoteles",
    goal="Encontrar alojamiento que encaje en el presupuesto restante",
    backstory="Conoces zonas y categorías de hotel en destinos europeos.",
)
hoteles_task = Task(
    description="Con el presupuesto restante tras los vuelos (ver paso anterior), propón 2 opciones de hotel para las fechas elegidas.",
    expected_output="2 opciones de hotel con precio por noche y zona, y presupuesto restante tras el hotel.",
    agent=hoteles_agent,
    context=[vuelos_task],
)

actividades_agent = Agent(
    role="Especialista en Actividades",
    goal="Proponer un plan de actividades dentro del presupuesto restante",
    backstory="Conoces atracciones, restaurantes y experiencias locales.",
)
actividades_task = Task(
    description=f"Con el presupuesto restante tras vuelos y hotel, propón un plan de actividades para los {dias} días.",
    expected_output="Un itinerario día por día con actividades y coste estimado, sin pasarse del presupuesto restante.",
    agent=actividades_agent,
    context=[vuelos_task, hoteles_task],
)

crew = Crew(
    agents=[vuelos_agent, hoteles_agent, actividades_agent],
    tasks=[vuelos_task, hoteles_task, actividades_task],
    process=Process.sequential,
    verbose=True,
)

print(crew.kickoff().raw)
```

**Lo que define el patrón aquí:** el orden es rígido a propósito — no tiene sentido planificar actividades antes de saber qué presupuesto sobra tras el hotel. Cada `context` apunta hacia atrás, nunca hacia delante.


## 2. Routing — clasificar el tipo de viajero primero

> **Clasificación: routing por código, no agéntico.** `elegir_planificador()` es una función Python con `if/elif` — ni siquiera hay un LLM decidiendo la categoría.

**Por qué este patrón aquí:** antes de planificar, conviene saber qué tipo de viaje es — un viaje de negocios de 2 días necesita un enfoque distinto a una escapada familiar de una semana. Routing elige al planificador adecuado según el tipo de consulta.


```python
from crewai import Agent, Task, Crew

planificador_business = Agent(
    role="Planificador de Viajes de Negocios",
    goal="Optimizar viajes cortos y eficientes para trabajo",
    backstory="Priorizas cercanía al centro, wifi fiable y traslados rápidos.",
)
planificador_familiar = Agent(
    role="Planificador de Viajes Familiares",
    goal="Planificar viajes cómodos para familias con niños",
    backstory="Priorizas actividades aptas para niños y ritmos relajados.",
)
planificador_aventura = Agent(
    role="Planificador de Viajes de Aventura",
    goal="Planificar viajes activos y poco convencionales",
    backstory="Priorizas experiencias únicas sobre la comodidad estándar.",
)

def elegir_planificador(consulta: str) -> Agent:
    q = consulta.lower()
    if any(w in q for w in ["negocio", "trabajo", "reunión", "conferencia"]):
        return planificador_business
    if any(w in q for w in ["niños", "familia", "hijos"]):
        return planificador_familiar
    if any(w in q for w in ["aventura", "trekking", "extremo"]):
        return planificador_aventura
    return planificador_familiar

consulta = "Viaje a Lisboa 4 días, 2 personas, 1200€, queremos relajarnos en pareja"
agente_elegido = elegir_planificador(consulta)

task = Task(
    description=f"Planifica este viaje: {consulta}",
    expected_output="Un itinerario de 4 días con vuelos, hotel y actividades, dentro de 1200€.",
    agent=agente_elegido,
)

crew = Crew(agents=[agente_elegido], tasks=[task], verbose=True)
print(crew.kickoff().raw)
```

**Lo que define el patrón aquí:** el routing ocurre *antes* de tocar vuelos/hoteles/actividades — decide qué "personalidad" de planificador va a resolver todo el problema, no qué especialista de viaje interviene.


## 3. Parallelization — los 4 especialistas a la vez

> **Clasificación: Workflow paralelo.** Yo decido en el código qué 4 tasks van en paralelo y cuál las combina; los agentes no negocian ese reparto entre sí.

**Por qué este patrón aquí:** vuelos, hoteles, actividades y transporte para un destino y fechas ya fijados son, en realidad, independientes entre sí — ninguno necesita el resultado de otro para empezar a buscar. Es el caso de manual de parallelization.


```python
from crewai import Agent, Task, Crew, Process

destino, dias, personas, presupuesto = "Lisboa", 4, 2, 1200

vuelos_agent = Agent(role="Especialista en Vuelos", goal="Buscar vuelos", backstory="Conoces tarifas aéreas.")
hoteles_agent = Agent(role="Especialista en Hoteles", goal="Buscar alojamiento", backstory="Conoces hoteles en Europa.")
actividades_agent = Agent(role="Especialista en Actividades", goal="Proponer actividades", backstory="Conoces qué hacer en cada ciudad.")
transporte_agent = Agent(role="Especialista en Transporte", goal="Planificar traslados locales", backstory="Conoces transporte público y privado en destinos europeos.")

vuelos_task = Task(
    description=f"Busca 2 opciones de vuelo a {destino} para {personas} personas.",
    expected_output="2 opciones de vuelo con precio aproximado.",
    agent=vuelos_agent,
    async_execution=True,
)
hoteles_task = Task(
    description=f"Busca 2 opciones de hotel en {destino} para {dias} días, {personas} personas.",
    expected_output="2 opciones de hotel con precio por noche.",
    agent=hoteles_agent,
    async_execution=True,
)
actividades_task = Task(
    description=f"Propón un plan de actividades en {destino} para {dias} días.",
    expected_output="Un itinerario día por día con actividades sugeridas.",
    agent=actividades_agent,
    async_execution=True,
)
transporte_task = Task(
    description=f"Propón cómo moverse dentro de {destino} durante {dias} días.",
    expected_output="Recomendación de transporte local con coste estimado.",
    agent=transporte_agent,
    async_execution=True,
)

coordinador = Agent(
    role="Coordinador de Itinerario",
    goal="Combinar propuestas de vuelos, hotel, actividades y transporte en un plan único",
    backstory="Ensamblas piezas sueltas en un itinerario coherente y dentro de presupuesto.",
)
combinar_task = Task(
    description=f"Combina las 4 propuestas anteriores en un itinerario único para {destino}, ajustando todo a {presupuesto}€ en total.",
    expected_output="Un itinerario final de 4 días con vuelos, hotel, actividades y transporte, y el coste total.",
    agent=coordinador,
    context=[vuelos_task, hoteles_task, actividades_task, transporte_task],
)

crew = Crew(
    agents=[vuelos_agent, hoteles_agent, actividades_agent, transporte_agent, coordinador],
    tasks=[vuelos_task, hoteles_task, actividades_task, transporte_task, combinar_task],
    # OJO: "Process.sequential" aquí NO significa "uno detrás de otro sin solapar".
    # CrewAI solo tiene 2 valores estables para `process`: sequential y hierarchical.
    # El paralelismo real lo dan los 4 `async_execution=True` de arriba, no este parámetro:
    # el motor recorre la LISTA de tasks en orden, pero lanza cada task async sin esperarla,
    # y solo bloquea en combinar_task porque su `context` depende de las 4 anteriores.
    process=Process.sequential,
    verbose=True,
)

print(crew.kickoff().raw)
```

**Aclaración importante sobre `Process.sequential` en este ejemplo:** el nombre puede confundir en un patrón que se llama "Parallelization". CrewAI solo tiene dos valores estables para `process`: `sequential` y `hierarchical` (un tercer valor, `consensual`, estuvo planeado pero fue retirado de la documentación oficial — no lo uses en código real). `Process.sequential` no significa "sin paralelismo": define cómo el motor recorre la *lista* de tasks, no si una task individual bloquea a las demás. El paralelismo real viene de `async_execution=True` en cada task — eso es lo que de verdad lanza las 4 búsquedas a la vez. `combinar_task` es la única que espera, y solo porque su `context` la obliga a hacerlo.

**Lo que define el patrón aquí:** las 4 búsquedas no se necesitan entre sí — solo la task final las necesita a *todas*. Comparado con Chaining (patrón 1), aquí nadie espera a nadie hasta el final.


## 4. Orchestrator-Workers — un agente de viajes que decide sobre la marcha

> **Clasificación: Multiagente real.** El manager LLM (`Process.hierarchical`) decide en tiempo de ejecución a quién delegar y cuánto — es el único ejemplo de los 7 donde esa decisión no está en mi código.

**Por qué este patrón aquí:** a diferencia de Parallelization, aquí no sabemos de antemano qué especialistas harán falta ni en qué proporción — depende de lo que pida el cliente. Un cliente que dice "solo quiero relajarme, no me importa el centro" necesita menos trabajo de Transporte y más de Actividades; uno con itinerario muy ajustado necesita más trabajo de Vuelos. El orquestador decide en tiempo real.


```python
from crewai import Agent, Task, Crew, Process

vuelos_agent = Agent(role="Especialista en Vuelos", goal="Buscar vuelos", backstory="Conoces tarifas aéreas.", allow_delegation=False)
hoteles_agent = Agent(role="Especialista en Hoteles", goal="Buscar alojamiento", backstory="Conoces hoteles en Europa.", allow_delegation=False)
actividades_agent = Agent(role="Especialista en Actividades", goal="Proponer actividades", backstory="Conoces qué hacer en cada ciudad.", allow_delegation=False)
transporte_agent = Agent(role="Especialista en Transporte", goal="Planificar traslados locales", backstory="Conoces transporte público y privado.", allow_delegation=False)

peticion_cliente = (
    "Viaje a Lisboa, 4 días, 2 personas, 1200€ en total. "
    "Lo único que de verdad importa es comer bien y ver el atardecer desde un buen mirador cada día; "
    "el resto (vuelos, hotel, cómo movernos) que sea lo más barato y simple posible."
)

main_task = Task(
    description=(
        f"Petición del cliente: {peticion_cliente}\n\n"
        "Decide qué especialistas necesitas y en qué orden, dado que las prioridades del "
        "cliente están claramente sesgadas hacia actividades y experiencia, no hacia logística. "
        "Entrega un itinerario completo y el presupuesto desglosado."
    ),
    expected_output="Un itinerario de 4 días con vuelos, hotel, actividades y transporte, reflejando las prioridades reales del cliente, dentro de 1200€.",
    agent=actividades_agent,
)

crew = Crew(
    agents=[vuelos_agent, hoteles_agent, actividades_agent, transporte_agent],
    tasks=[main_task],
    process=Process.hierarchical,
    manager_llm="gpt-4o-mini",
    verbose=True,
)

print(crew.kickoff().raw)
```

**Lo que define el patrón aquí:** el manager (`Process.hierarchical`) lee la petición y decide cuánto delega a cada especialista — quizá apenas consulta a Transporte, y vuelve dos veces a Actividades. Esa asignación dinámica es justo lo que Parallelization no puede hacer, porque ahí las 4 tasks ya están fijadas antes de arrancar.


## 5. Evaluator-Optimizer — el itinerario tiene que cuadrar con el presupuesto

> **Clasificación: ni multiagente ni multi-paso.** Es un único agente con un bucle de regeneración; el "evaluador" es una función Python, no otro LLM.

**Por qué este patrón aquí:** es habitual que un primer borrador de itinerario se pase de presupuesto. En vez de confiar en que el agente lo calcule bien a la primera, un evaluador revisa el total y obliga a recortar hasta que cuadre.


```python
from crewai import Agent, Task, Crew
from crewai.tasks.task_output import TaskOutput
import re

PRESUPUESTO_MAXIMO = 1200

planificador = Agent(
    role="Planificador de Viajes",
    goal="Crear itinerarios completos a Lisboa dentro de presupuesto",
    backstory="Diseñas itinerarios de 4 días con vuelos, hotel y actividades incluidos.",
)

def validar_presupuesto(output: TaskOutput):
    """Busca un total en euros en el texto y comprueba que no supere el máximo."""
    match = re.search(r"total[^\d]*(\d+)\s*€", output.raw, re.IGNORECASE)
    if not match:
        return (False, "No encuentro un total claro en euros. Indica el coste total del itinerario explícitamente como 'Total: X€'.")
    total = int(match.group(1))
    if total > PRESUPUESTO_MAXIMO:
        return (False, f"El itinerario cuesta {total}€, supera el presupuesto de {PRESUPUESTO_MAXIMO}€ por {total - PRESUPUESTO_MAXIMO}€. Recorta gastos (hotel más barato, menos actividades de pago) y vuelve a calcular el total.")
    return (True, output.raw)

itinerario_task = Task(
    description=(
        f"Crea un itinerario de 4 días a Lisboa para 2 personas: vuelos, hotel y actividades. "
        f"Indica el coste total al final como 'Total: X€'. Presupuesto máximo: {PRESUPUESTO_MAXIMO}€."
    ),
    expected_output="Itinerario de 4 días con vuelos, hotel, actividades, y una línea final 'Total: X€'.",
    agent=planificador,
    guardrail=validar_presupuesto,
    guardrail_max_retries=4,
)

crew = Crew(agents=[planificador], tasks=[itinerario_task], verbose=True)
print(crew.kickoff().raw)
```

**Lo que define el patrón aquí:** el propio planificador genera y regenera — el evaluador no es otro especialista de viaje, es una función que solo sabe sumar y comparar contra el límite. Eso es justo lo que distingue Evaluator-Optimizer: el criterio de calidad es externo y objetivo, no otra opinión de dominio.


## 6. Jerárquico — agencia central + delegaciones por destino

> **Clasificación: Workflow que orquesta varios crews.** Yo ejecuto cada `Crew` por separado en Python y paso el resultado a mano (`reporte_europa`, `reporte_actividades`) al siguiente nivel — los crews no se llaman entre sí solos.

**Por qué este patrón aquí:** imagina una agencia de viajes con una oficina central y delegaciones especializadas por región (Europa, América). Cada delegación coordina sus propios especialistas de vuelos/hoteles antes de que la central decida el plan final — dos niveles de coordinación, no solo uno.


```python
from crewai import Agent, Task, Crew, Process

vuelos_europa = Agent(role="Vuelos Europa", goal="Buscar vuelos dentro de Europa", backstory="Conoces aerolíneas europeas low-cost y tradicionales.")
hoteles_europa = Agent(role="Hoteles Europa", goal="Buscar hoteles en ciudades europeas", backstory="Conoces alojamiento en Lisboa, Roma, París.")

vuelos_task_eu = Task(
    description="Busca 2 opciones de vuelo a Lisboa para 2 personas, 4 días.",
    expected_output="2 opciones de vuelo con precio.",
    agent=vuelos_europa,
)
hoteles_task_eu = Task(
    description="Busca 2 opciones de hotel en Lisboa para 4 noches, 2 personas.",
    expected_output="2 opciones de hotel con precio por noche.",
    agent=hoteles_europa,
)
crew_delegacion_europa = Crew(
    agents=[vuelos_europa, hoteles_europa],
    tasks=[vuelos_task_eu, hoteles_task_eu],
    process=Process.sequential,
)

actividades_agent = Agent(role="Actividades Lisboa", goal="Proponer actividades en Lisboa", backstory="Conoces Lisboa al detalle.")
transporte_agent = Agent(role="Transporte Lisboa", goal="Planificar movilidad en Lisboa", backstory="Conoces metro, tranvías y taxis de Lisboa.")

actividades_task = Task(
    description="Propón un plan de actividades de 4 días en Lisboa.",
    expected_output="Itinerario de actividades día por día.",
    agent=actividades_agent,
)
transporte_task = Task(
    description="Propón cómo moverse en Lisboa durante 4 días.",
    expected_output="Recomendación de transporte local con coste estimado.",
    agent=transporte_agent,
)
crew_delegacion_actividades = Crew(
    agents=[actividades_agent, transporte_agent],
    tasks=[actividades_task, transporte_task],
    process=Process.sequential,
)

reporte_europa = crew_delegacion_europa.kickoff().raw
reporte_actividades = crew_delegacion_actividades.kickoff().raw

director_agencia = Agent(
    role="Director de Agencia de Viajes",
    goal="Ensamblar el itinerario final a partir de los reportes de cada delegación",
    backstory="Coordinas todas las delegaciones y entregas el itinerario final al cliente.",
)
itinerario_final_task = Task(
    description=(
        f"Reporte de vuelos y hotel:\n{reporte_europa}\n\n"
        f"Reporte de actividades y transporte:\n{reporte_actividades}\n\n"
        "Ensambla el itinerario final completo para el cliente, con el coste total, dentro de 1200€."
    ),
    expected_output="Itinerario final de 4 días con todo incluido y coste total.",
    agent=director_agencia,
)
crew_central = Crew(agents=[director_agencia], tasks=[itinerario_final_task], process=Process.sequential)

print(crew_central.kickoff().raw)
```

**Lo que define el patrón aquí:** hay dos niveles de coordinación reales — cada delegación resuelve su propia parte de forma autónoma (nivel táctico) antes de que la central (nivel estratégico) vea nada. Compáralo con Orchestrator-Workers (patrón 4): ahí hay un solo nivel de mando sobre 4 especialistas sueltos; aquí hay un nivel intermedio que ya agrupa especialistas antes de subir.


## 7. Democrático — la pareja no se pone de acuerdo en el destino

> **Clasificación: multiagente débil.** Cada agente opina de forma independiente (eso sí es autonomía real), pero el escrutinio final sigue una receta fija mía, no una negociación dinámica entre los agentes.

**Por qué este patrón aquí:** antes de planificar nada, dos personas con preferencias distintas tienen que decidir juntas qué tipo de viaje quieren. No hay un jefe que decida — se vota.

**Nota sobre `Process.consensual`:** si buscas en foros o tutoriales antiguos, verás referencias a un tercer valor de `process` llamado `consensual`, pensado exactamente para este patrón. Estuvo planeado, pero **fue retirado de la documentación oficial de CrewAI** y no es estable — no lo uses en código real. Por eso este ejemplo simula la votación a mano con tasks y un escrutinio, igual que en el documento anterior.


```python
from crewai import Agent, Task, Crew, Process

opciones_viaje = [
    "Lisboa: cultura, gastronomía, ritmo tranquilo",
    "Ámsterdam: museos, bicicleta, vida nocturna",
    "Praga: histórico, económico, cerveza artesanal",
]

viajero_1 = Agent(
    role="Preferencias de Viajero 1",
    goal="Evaluar destinos según el gusto por la gastronomía y la tranquilidad",
    backstory="Para ti, lo importante es comer bien y no correr.",
)
viajero_2 = Agent(
    role="Preferencias de Viajero 2",
    goal="Evaluar destinos según el gusto por la cultura y el presupuesto",
    backstory="Para ti, lo importante es aprender cosas nuevas sin gastar de más.",
)
amigo_consultado = Agent(
    role="Amigo que ya viajó a los 3 destinos",
    goal="Dar una opinión informada basada en experiencia previa",
    backstory="Has estado en los 3 destinos y tienes opiniones formadas.",
)

votantes = [viajero_1, viajero_2, amigo_consultado]
vote_tasks = []
for votante in votantes:
    t = Task(
        description=(
            f"De estas 3 opciones de destino:\n" + "\n".join(opciones_viaje) +
            "\n\nElige UNA y justifica tu elección en 1-2 frases, según tu propia perspectiva."
        ),
        expected_output="El destino elegido (solo el nombre de la ciudad) y una justificación breve.",
        agent=votante,
    )
    vote_tasks.append(t)

escrutador = Agent(
    role="Escrutador",
    goal="Contar los votos y anunciar el destino ganador sin opinar",
    backstory="Tu único trabajo es contar votos con exactitud y sin favoritismo.",
)
escrutinio_task = Task(
    description="Cuenta los votos de las 3 evaluaciones anteriores y anuncia qué destino ganó por mayoría (o si hay empate).",
    expected_output="Recuento de votos por destino, y el destino ganador (o empate, con los destinos empatados).",
    agent=escrutador,
    context=vote_tasks,
)

crew = Crew(
    agents=votantes + [escrutador],
    tasks=vote_tasks + [escrutinio_task],
    process=Process.sequential,
    verbose=True,
)

print(crew.kickoff().raw)
```

**Lo que define el patrón aquí:** la decisión de *qué planificar* — no cómo planificarlo — se reparte entre 3 voces sin jerarquía. Una vez hay un destino ganador, podrías encadenar cualquiera de los 6 patrones anteriores para planificar el viaje en sí.


## Comparación final: el mismo viaje, 7 formas de organizarlo

| Patrón | ¿Workflow o multiagente? | Qué decide y cuándo | Especialistas de viaje implicados |
|---|---|---|---|
| Chaining | Workflow | El orden de pasos es fijo: vuelos → hotel → actividades | Vuelos, Hoteles, Actividades (en ese orden) |
| Routing | Routing por código (ni agéntico) | Se elige el *tipo* de planificador antes de planificar nada | Ninguno aún — decide quién planifica |
| Parallelization | Workflow paralelo | Las 4 búsquedas corren a la vez, se combinan al final | Vuelos, Hoteles, Actividades, Transporte (en paralelo) |
| Orchestrator-Workers | **Multiagente real** | Un manager LLM decide cuánto peso dar a cada especialista, según la petición | Los mismos 4, pero con reparto dinámico |
| Evaluator-Optimizer | Ni una cosa ni otra | El itinerario se regenera hasta caber en presupuesto | Un único planificador, evaluado por una función de coste |
| Jerárquico | Workflow (orquesta varios crews) | Delegaciones resuelven su parte antes de subir a la central | Vuelos+Hoteles (delegación 1), Actividades+Transporte (delegación 2) |
| Democrático | Multiagente débil | Se vota el destino antes de planificar | Ninguno aún — decide qué se va a planificar |

El mismo dominio, 7 arquitecturas: la diferencia entre patrones no está en *qué* hace cada agente de viaje, sino en *quién decide el orden y el reparto del trabajo*. Y como muestra la columna nueva, en la mayoría de estos 7 casos quien decide sigue siendo tu código — solo en Orchestrator-Workers se lo cedes de verdad a un LLM.
